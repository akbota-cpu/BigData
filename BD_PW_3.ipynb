```python
import numpy as np
# Сатып алу сомалары
data = np.array([
    5000, 7000, 8000, 6000, 9000,
    7500, 8500, 6500, 10000, 500000
])
# Орташа мән
mean = np.mean(data)
# Стандартты ауытқу
std = np.std(data)
# Z-score есептеу
z_scores = (data - mean) / std
# Аномалияларды анықтау
anomalies = data[np.abs(z_scores) > 2]
print("Орташа мән:", mean)
print("Z-score:", z_scores)
print("Аномалиялар:", anomalies)
```

```python
anomalies = data[np.abs(z_scores) > 2]
print("Анықталған аномалиялар:", anomalies)
```

```python
indisec = np.where(np.abs(z_scores) > 2)[0]
print("Аномалия индексі:", indisec)
print("Аномалия мәндері:", data[indisec])
```

```python
# Аномалияларды алып тастау
clean_data = data[np.abs(z_scores) <= 2]

# Жаңа орташа мән
new_mean = np.mean(clean_data)

print("Бастапқы орташа мән:", np.mean(data))
print("Аномалиясыз орташа мән:", new_mean)
```

```python
print("Аномалияға дейінгі орташа мән:", np.mean(data))
print("Аномалиядан кейінгі орташа мән:", np.mean(clean_data))
```

```python
import numpy as np

np.random.seed(42)
sales = np.random.randint(
    1000,
    50000,
    size=(4, 12, 30)
)
print("Массивтің өлшемі:", sales.shape)

# 1. Бірінші аймақтың барлық деректерін алыңыз
region1 = sales[0]
print("Бірінші аймақтың деректері:\n", region1)
```

```python
# 2. Үшінші аймақтың 5-айдағы барлық күндік сатылымдарын алыңыз
region3_month5 = sales[2, 4]
print("Үшінші аймақтың 5-айдағы сатылымдары:\n", region3_month5)
```

```python
# 3. Барлық аймақтардың бірінші айдағы сатылымдарын алыңыз
month1_all_regions = sales[:, 0, :]
print("Барлық аймақтардың бірінші айдағы сатылымдары:\n", month1_all_regions)
```

```python
# 4. Бірінші және екінші аймақтардың деректерін бөліп алыңыз
region1_2 = sales[0:2]
print("Бірінші және екінші аймақтардың деректері (өлшемі):", region1_2.shape)
```

```python
# 5. Әр аймақ бойынша жылдық сатылым көлемін есептеңіз
yearly_sales_per_region = sales.sum(axis=(1, 2))
print("Әр аймақтың жылдық сатылым көлемі:", yearly_sales_per_region)
```

```python
# 6. Әр ай бойынша барлық аймақтардың сатылымын есептеңіз
monthly_sales_all_regions = sales.sum(axis=(0, 2))
print("Әр ай бойынша барлық аймақтардың сатылымы:", monthly_sales_all_regions)
```

```python
import numpy as np

np.random.seed(42)
n = 1_000_000

# 1-КЕЗЕҢ. ДЕРЕКТЕРДІ ҚАЛЫПТАСТЫРУ

customer_id = np.random.randint(1, 50001, n).astype(float)      # Клиент ID
product_id = np.random.randint(1, 1001, n).astype(float)        # Өнім ID
purchase_amount = np.random.uniform(100, 50000, n)              # Сатып алу сомасы
quantity = np.random.randint(1, 11, n).astype(float)            # Тауар саны
discount = np.random.uniform(0, 0.5, n)                         # Жеңілдік мөлшері (0-50%)
region_code = np.random.randint(1, 6, n).astype(float)          # Аймақ коды (1-5)

data = np.column_stack([
    customer_id, product_id, purchase_amount, quantity, discount, region_code
])

columns = ["customer_id", "product_id", "purchase_amount", "quantity", "discount", "region_code"]

print("Массивтің өлшемі:", data.shape)
print("Массивтің dtype:", data.dtype)
```

```python
# АҚАУЛАРДЫ ЖАСАНДЫ ЕНГІЗУ

# Қайталанатын жазбалар
dup_idx = np.random.choice(n, 5000, replace=False)
data = np.vstack([data, data[dup_idx]])

# Жетіспейтін мәндер (purchase_amount бағанына NaN)
nan_idx = np.random.choice(data.shape[0], 3000, replace=False)
data[nan_idx, 2] = np.nan

# Аномалиялар (шектен тыс үлкен сома)
outlier_idx = np.random.choice(data.shape[0], 500, replace=False)
data[outlier_idx, 2] = np.random.uniform(500000, 2000000, 500)


# 2-КЕЗЕҢ. ДЕРЕКТЕРДІ ТЕКСЕРУ

unique_rows, counts = np.unique(data, axis=0, return_counts=True)
print("Артық (қайталанған) жолдар саны:", data.shape[0] - unique_rows.shape[0])

missing_mask = np.isnan(data[:, 2])
print("Жетіспейтін мәндер саны:", missing_mask.sum())

valid_amounts = data[~missing_mask, 2]
mean_amt = np.mean(valid_amounts)
std_amt = np.std(valid_amounts)
z_scores = np.abs((data[:, 2] - mean_amt) / std_amt)
outlier_mask = z_scores > 3
print("Аномалиялар саны (z-score > 3):", np.nansum(outlier_mask))


# 3-КЕЗЕҢ. ДЕРЕКТЕРДІ ӨҢДЕУ

median_amt = np.nanmedian(data[:, 2])
data[missing_mask, 2] = median_amt                 # жетіспейтін мәндерді толтыру

data_unique = np.unique(data, axis=0)               # қайталануларды жою

z_scores_clean = np.abs((data_unique[:, 2] - np.mean(data_unique[:, 2])) / np.std(data_unique[:, 2]))
clean_data = data_unique[z_scores_clean <= 3]        # аномалияларды жою

selected_columns = clean_data[:, [0, 2, 5]]          # қажетті бағандарды бөліп алу
filtered = clean_data[clean_data[:, 2] > 10000]      # шарт бойынша фильтр


# 4-КЕЗЕҢ. МАТЕМАТИКАЛЫҚ ЕСЕПТЕУЛЕР

total_revenue = np.sum(clean_data[:, 2])
print("Жалпы табыс:", round(total_revenue, 2))

region_sales = {}
for region in np.unique(clean_data[:, 5]):
    region_sales[int(region)] = np.sum(clean_data[clean_data[:, 5] == region, 2])
print("Әр аймақтың жалпы сатылымы:", region_sales)


# 5-КЕЗЕҢ. СТАТИСТИКАЛЫҚ ТАЛДАУ

amounts = clean_data[:, 2]
print("Орташа мән:", round(np.mean(amounts), 2))
print("Медиана:", round(np.median(amounts), 2))
print("Минимум:", round(np.min(amounts), 2))
print("Максимум:", round(np.max(amounts), 2))
print("Стандартты ауытқу:", round(np.std(amounts), 2))
print("Q1:", round(np.percentile(amounts, 25), 2))
print("Q3:", round(np.percentile(amounts, 75), 2))
```
